# Brew & Bean Data Cleaning

## Objective

Clean and prepare the generated datasets for exploratory analysis and Power BI.

Datasets:
- Branches
- Products
- Employees
- Customers
- Transactions

Output:
Clean datasets saved into the `data/processed` folder.

```python
import pandas as pd
from pathlib import Path
```

```python
DATA_DIR = Path("../data")

GENERATED_DIR = DATA_DIR / "generated"
PROCESSED_DIR = DATA_DIR / "processed"
```

```python
branches_df = pd.read_csv(GENERATED_DIR / "branches.csv")
products_df = pd.read_csv(GENERATED_DIR / "products.csv")
employees_df = pd.read_csv(GENERATED_DIR / "employees.csv")
customers_df = pd.read_csv(GENERATED_DIR / "customers.csv")
transactions_df = pd.read_csv(GENERATED_DIR / "transactions.csv")
```

```python
datasets = {
    "Branches": branches_df,
    "Products": products_df,
    "Employees": employees_df,
    "Customers": customers_df,
    "Transactions": transactions_df
}

for name, df in datasets.items():
    print("=" * 50)
    print(name)
    print("=" * 50)
    print(df.head())
    print()
```

```python
for name, df in datasets.items():
    print("=" * 50)
    print(name)
    print("=" * 50)
    print(df.info())
    duplicates = df.duplicated().sum()

    print(f"Duplicate Rows: {duplicates}")
    print()
```

```python
employees_df["Hire_Date"] = pd.to_datetime(employees_df["Hire_Date"])
customers_df["Join_Date"] = pd.to_datetime(customers_df["Join_Date"])
transactions_df["Transaction_Date"] = pd.to_datetime(
    transactions_df["Transaction_Date"]
)
```

```python
for name, df in datasets.items():
    print("=" * 50)
    print(name)
    print("=" * 50)
    print(df.dtypes)
    print()
```

```python
for name, df in datasets.items():
    print("=" * 50)
    print(name)
    print("=" * 50)

    print(df.isnull().sum())
    print()
```

```python
primary_keys = {
    "Branches": ("Branch_ID", branches_df),
    "Products": ("Product_ID", products_df),
    "Employees": ("Employee_ID", employees_df),
    "Customers": ("Customer_ID", customers_df),
    "Transactions": ("Transaction_ID", transactions_df)
}

for table, (key, df) in primary_keys.items():
    print(f"{table}")
    print(f"{key} is unique: {df[key].is_unique}")
    print()
```

```python
branches_df.to_csv(PROCESSED_DIR / "branches.csv", index=False)
products_df.to_csv(PROCESSED_DIR / "products.csv", index=False)
employees_df.to_csv(PROCESSED_DIR / "employees.csv", index=False)
customers_df.to_csv(PROCESSED_DIR / "customers.csv", index=False)
transactions_df.to_csv(PROCESSED_DIR / "transactions.csv", index=False)

print("✅ All cleaned datasets have been saved to data/processed/")
```
