## Employee Data
Generate realistic employee records for each Brew & Bean branch.

```python
job_roles = {
    "Branch Manager": 1,
    "Supervisor": 1,
    "Barista": 3,
    "Cashier": 2,
    "Kitchen Staff": 2
}
```

```python
employees = []

employee_number = 1

for _, branch in branches_df.iterrows():

    for role, count in job_roles.items():

        for _ in range(count):

            employees.append({

                "Employee_ID": f"E{employee_number:04d}",

                "Employee_Name": fake.name(),

                "Branch_ID": branch["Branch_ID"],

                "Branch_Name": branch["Branch_Name"],

                "Job_Role": role,

                "Hire_Date": fake.date_between(
                    start_date="-6y",
                    end_date="today"
                ),

                "Monthly_Salary_KES": {
                    "Branch Manager": 95000,
                    "Supervisor": 70000,
                    "Barista": 38000,
                    "Cashier": 35000,
                    "Kitchen Staff": 34000
                }[role]

            })

            employee_number += 1
```

```python
employees_df = pd.DataFrame(employees)

employees_df.head()
```

```python
employees_df.to_csv(
    GENERATED_DIR / "employees.csv",
    index=False
)

print("✅ employees.csv saved successfully!")
```

## Customer Data

Generate realistic Brew & Bean customers for loyalty and purchasing analysis.

```python
NUM_CUSTOMERS = 500
```

random.choice() picks one item with equal chance.
random.choices() lets you assign weights.

```python
customers = []

for i in range(1, NUM_CUSTOMERS + 1):

    customer_type = random.choices(
        population=[
            "Student",
            "Professional",
            "Remote Worker",
            "Tourist",
            "Regular"
        ],
        weights=[30, 35, 15, 5, 15],
        k=1
    )[0]

    customers.append({

        "Customer_ID": f"C{i:05d}",

        "Customer_Name": fake.name(),

        "Age": random.randint(18, 65),

        "Gender": random.choice([
            "Male",
            "Female"
        ]),

        "Customer_Type": customer_type,

        "Preferred_Branch": random.choice(
            BRANCHES
        ),

        "Loyalty_Member": random.choices(
            ["Yes", "No"],
            weights=[65, 35],
            k=1
        )[0],

        "Join_Date": fake.date_between(
            start_date="-5y",
            end_date="today"
        )

    })
```

```python
customers_df = pd.DataFrame(customers)

customers_df.head()
```

```python
customers_df.to_csv(
    GENERATED_DIR / "customers.csv",
    index=False
)

print("✅ customers.csv saved successfully!")
```

```python
customers_df.info()
customers_df.describe(include="all")
```

## Sales Transactions

Generate realistic sales transactions across all Brew & Bean branches.

```python
NUM_TRANSACTIONS = 50000
```

```python
transactions = []

for i in range(1, NUM_TRANSACTIONS + 1):

    customer = customers_df.sample(1).iloc[0]
    product = products_df.sample(1).iloc[0]
    employee = employees_df.sample(1).iloc[0]

    quantity = random.randint(1, 3)

    transactions.append({

        "Transaction_ID": f"T{i:06d}",

        "Transaction_Date": fake.date_between(
            start_date="-2y",
            end_date="today"
        ),

        "Customer_ID": customer["Customer_ID"],

        "Branch_ID": employee["Branch_ID"],

        "Employee_ID": employee["Employee_ID"],

        "Product_ID": product["Product_ID"],

        "Quantity": quantity,

        "Unit_Price": product["Price_KES"],

        "Total_Amount": quantity * product["Price_KES"],

        "Payment_Method": random.choice(PAYMENT_METHODS)

    })
```

```python
transactions_df = pd.DataFrame(transactions)

transactions_df.head()
```

```python
transactions_df.to_csv(
    GENERATED_DIR / "transactions.csv",
    index=False
)

print("✅ transactions.csv saved successfully!")
```
